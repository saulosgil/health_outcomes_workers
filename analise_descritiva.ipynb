# Análise descritiva para confecção da tabela 1.

```python
import pandas as pd
import numpy as np

caminho = r'C:\Users\saulosgil\OneDrive\Documentos\health_outcomes_workers\data\df_analise.csv'
df = pd.read_csv(caminho)

df.head()
```

## Calcular N (%) de cada valor da variável sexo

```python
tabela = pd.DataFrame({
    'N': df['sexo'].value_counts().sort_index(),
    '%': df['sexo'].value_counts(normalize=True).sort_index().mul(100).round(0),
})
tabela
```

## Criar variável `idade_cat` e calcular N (%) de cada valor

 - idade < 65;
 
 - idade >= 65.

```python
idade = df['idade'].astype(int)

# Binária: < 65 = 1, >= 65 = 0, ausente = NaN
df['idade_cat'] = np.where(idade < 65, '< 65 anos', '>= 65 anos')

tabela = pd.DataFrame({
    'N': df['idade_cat'].value_counts().sort_index(),
    '%': df['idade_cat'].value_counts(normalize=True).sort_index().mul(100).round(0),
})
tabela
```

## Calcular N (%) de cada valor da variável `cor_raca`

```python
tabela = pd.DataFrame({
    'N': df['cor_raca'].value_counts().sort_index(),
    '%': df['cor_raca'].value_counts(normalize=True).sort_index().mul(100).round(0),
})
tabela
```

## Calcular N (%) de cada valor da variável `estado_civil`

```python
# juntar separado e divorciado
df['estado_civil'] = df['estado_civil'].replace({'divorciado': 'separado'})

tabela = pd.DataFrame({
    'N': df['estado_civil'].value_counts().sort_index(),
    '%': df['estado_civil'].value_counts(normalize=True).sort_index().mul(100).round(0),
})
tabela
```

## Calcular N (%) de cada valor da variável `escolaridade`

```python
mapa_escolaridade = {
    'não estudou':              'No formal education',
    'fundamental incompleto':   'Incomplete elementary school',
    'fundamental completo':     'Complete elementary school',
    'médio incompleto':         'Incomplete high school',
    'médio completo':           'High school graduate',
    'superior incompleto':      'Some college',
    'superior completo':        "College degree",
}

ordem = list(mapa_escolaridade.values())

df['escolaridade_en'] = pd.Categorical(
    df['escolaridade'].map(mapa_escolaridade),
    categories=ordem,
    ordered=True,
)

tabela = pd.DataFrame({
    'N': df['escolaridade_en'].value_counts().sort_index(),
    '%': df['escolaridade_en'].value_counts(normalize=True).sort_index().mul(100).round(0),
})
tabela
```
