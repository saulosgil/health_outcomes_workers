## Tratamento dos valores ausentes

```python
# 1. Separar apenas as colunas numéricas
num_cols = df.select_dtypes(include='number').columns

# 2. Separar apenas as colunas de texto
str_cols = df.select_dtypes(include='str').columns

# 3. Preencher os NaN com a média de cada coluna (no próprio df)
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())

# 4. Preencher os NaN com a moda de cada coluna (no próprio df)
df[str_cols] = df[str_cols].fillna(df[str_cols].mode().iloc[0])

print("Quantidade de valores ausentes por coluna depois da imputação:")
print(df.isna().sum())
print(80 *'-')
```

## Tratar coluna diagnóstico parar criar as variáveis DM2 e HAS. Tratar a coluna imc_classe para pegar a condição obesidade.

```python

# Criação de novas colunas para indicar a presença de DM, HAS e obesidade
df_selecionada['DM2'] = (
    df_selecionada['diagnostico']
    .str.contains('DM', case=False, na=False)
    .astype(int)          # 1 = tem DM, 0 = não tem
)

df_selecionada['HAS'] = (
    df_selecionada['diagnostico']
    .str.contains('HAS', case=False, na=False)
    .astype(int)          # 1 = tem HAS, 0 = não tem
)

df_selecionada['obesidade'] = (
    df_selecionada['imc_classe']
    .str.contains('obesidade', case=False, na=False)
    .astype(int)          # 1 = tem obesidade, 0 = não tem
)

# Remover as colunas 'diagnostico' e 'imc_classe' do dataframe df_selecionada
df_selecionada = df_selecionada.drop(columns=['diagnostico', 'imc_classe'])

df_selecionada.head()
```

Agora temos as comorbidades (DM2, HAS e obesidade) como variáveis binárias (0 = não tem, 1 = tem).

```python
df_selecionada['medicamentos'].unique()
```

```python
# Tratar a coluna medicamentos para criar uma coluna binária indicando se a pessoa faz uso de medicamentos ou não
# df_selecionada['medicamentos'].unique() # descomentar apenas para ver a lista completa de medicamentos

padroes = {
    'ARB':          r'sartan',                                   # losartana, valsartana, olmesartana...
    'Diuretics':    r'diur[eé]tic|tiazid|clortalidona|furosemida|espironolactona|indapamida',
    'Beta_blockers':r'olol',                                     # atenolol, propranolol, metoprolol, carvedilol...
    'ACE_inhibitors': r'pril',                                   # enalapril, captopril, lisinopril, ramipril...
    'Metformin':    r'metformin|glifage',
    'CCB':          r'dipin|diltiazem|verapamil',                # anlodipino, nifedipino...
}

col = df_selecionada['medicamentos'].str.lower()

for classe, padrao in padroes.items():
    df_selecionada[classe] = col.str.contains(padrao, na=False, regex=True).astype(int)

# Remover coluna medicamentos do dataframe df_selecionada
df_selecionada = df_selecionada.drop(columns=['medicamentos'])

df_selecionada.head()
```

# Criar a coluna aumentado risco para doenças cardio metabólicas

Para essa classificação será utilizada a relação cintura‑altura;

A relação cintura‑altura é calculada dividindo a circunferência da cintura pela altura; para adultos, manter o valor abaixo de 0,50 indica menor risco cardiometabólico segundo diretrizes como NICE e estudos de Ashwell & Gibson

< 0,40 — abaixo do adequado
0,40 a 0,49 — saudável
0,50 a 0,59 — risco aumentado
≥ 0,60 — risco elevado

```python
df_selecionada['sexo'].unique()
```

```python
# Seleciona a coluna 'cintura_altura_ratio' e arredonda os valores para 2 casas decimais
# Converte a razão de porcentagem para proporção (ex.: 52.23 → 0.5223)
rce = df_selecionada['cintura_altura_ratio'].astype(float) / 100

# Binária: ≥ 0,50 = 1, < 0,50 = 0, ausente = NaN
df_selecionada['aumentado_cmr'] = np.where(rce.isna(), np.nan, (rce >= 0.50).astype(int))

# Move a coluna para logo depois de 'cintura_altura_ratio'
col = df_selecionada.pop('aumentado_cmr')
pos = df_selecionada.columns.get_loc('cintura_altura_ratio') + 1
df_selecionada.insert(pos, 'aumentado_cmr', col)

df_selecionada[['cintura_altura_ratio', 'aumentado_cmr']].head(10)
```

# Calcular WHOQoL-BREF

```python
who = df_selecionada.filter(regex='^who').astype(float).copy()

# 1. Valores fora de 1–5 viram ausentes (ELSE=SYSMIS)
who = who.where(who.isin([1, 2, 3, 4, 5]))

# 2. Inverter os itens 3, 4 e 26 (1=5, 2=4, 3=3, 4=2, 5=1)
for i in [3, 4, 26]:
    who[f'who{i}'] = 6 - who[f'who{i}']

# 3. Escores dos domínios (MEAN.n = média só se houver pelo menos n itens válidos) × 4
def media_min(df, itens, n_min):
    cols = [f'who{i}' for i in itens]
    media = df[cols].mean(axis=1)
    validos = df[cols].notna().sum(axis=1)
    return media.where(validos >= n_min) * 4

dominios = {
    'whoqol_fisico':      ([3, 4, 10, 15, 16, 17, 18], 6),           
    'whoqol_psicologico': ([5, 6, 7, 11, 19, 26], 5),                
    'whoqol_social':      ([20, 21, 22], 2),                        
    'whoqol_ambiente':    ([8, 9, 12, 13, 14, 23, 24, 25], 6),
}

for nome, (itens, n_min) in dominios.items():
    escore_4_20 = media_min(who, itens, n_min)
    # 4. Transformar para escala 0–100
    df_selecionada[nome] = round((escore_4_20 - 4) * (100 / 16), 2)

# 5. Excluir quem respondeu menos de 21 dos 26 itens (> 20% ausentes)
total_validos = who.notna().sum(axis=1)
df_selecionada['whoqol_n_validos'] = total_validos
df_selecionada.loc[total_validos < 21, list(dominios)] = np.nan

# remove as questões dos who e deixa apenas os dominios calculados
itens_who = df_selecionada.filter(regex=r'^who\d+$').columns
df_selecionada = df_selecionada.drop(columns=itens_who)
df_selecionada.filter(regex='^who').columns.tolist() # verifica quais colunas ficaram
```
